XXVII. Znajdź drugą największą liczbę w liście:
[10, 5, 8, 20, 3, 15]


```python
lst = [10, 5, 8, 20, 3, 15]

biggest = max(lst[0], lst[1])
second_biggest = min(lst[0], lst[1])

for i in lst[2:]:
    if i > biggest:
        second_biggest, biggest = biggest, i
    elif i > second_biggest and i != biggest:
        second_biggest = i

print(second_biggest)
```

XXVIII. Zamień wszystkie małe litery w tekście na wielkie używając pętli (bez używania funkcji upper()).

```python
text = "sdfghjASDFG"
text_upper = ""

for i in text:
    code = ord(i)

    if 97 <= code <= 122:
        text_upper += chr(code - 32)
    else:
        text_upper += i

print(text_upper)
```

# Pętla while
I. Wczytuj liczby od użytkownika dopóki użytkownik nie wpisze 0. Następnie wypisz sumę wszystkich wprowadzonych liczb.


```python
user_input = 1
total_sum = 0

while user_input != 0:
    user_input = int(input("Podaj liczbę (0 kończy program): "))
    total_sum += user_input

print(total_sum)
```

II. Wczytuj liczby od użytkownika dopóki nie zostanie wprowadzona liczba ujemna. Na końcu wypisz ile liczb zostało wprowadzonych.

```python
user_input = 0
count = 0

while user_input >= 0:
    user_input = int(input("Podaj liczbę (ujemna kończy): "))

    if user_input >= 0:
        count += 1

print(count)
```

III. Wczytuj liczby od użytkownika dopóki użytkownik nie wpisze 0. Jeśli pętla zakończy się normalnie (bez użycia break), wypisz komunikat "End of data input" w bloku else.

```python
num = 1

while num != 0:
    num = int(input("Podaj liczbę (0 kończy): "))
else:
    print("End of data input")
```

IV. Wczytuj liczby od użytkownika i oblicz ich sumę. Jeśli użytkownik wpisze liczbę ujemną, zakończ pętlę używając instrukcji break.


```python
total_sum = 0

while True:
    user_input = int(input("Podaj liczbę (ujemna kończy): "))

    if user_input < 0:
        break

    total_sum += user_input

print(total_sum)
```

V. Sprawdź czy podana liczba jest liczbą pierwszą używając pętli while. Jeśli podczas sprawdzania znajdziesz dzielnik użyj break. Jeśli pętla zakończy się bez przerwania wypisz w bloku else, że liczba jest pierwsza.


```python
num = 17

if num < 2:
    print("Liczba nie jest pierwsza")
else:
    i = 2

    while i * i <= num:
        if num % i == 0:
            print("Liczba nie jest pierwsza")
            break

        i += 1
    else:
        print("Liczba jest pierwsza")
```

VI. Przeglądaj elementy listy używając pętli while i sprawdź czy liczba 10 znajduje się w liście. Jeśli zostanie znaleziona użyj break. Jeśli pętla zakończy się bez przerwania wypisz "Number not found" w bloku else.
Lista: [3, 7, 4, 8, 2]


```python
lst = [3, 7, 4, 8, 2]
ind = 0

while ind < len(lst):
    if lst[ind] == 10:
        print("jest 10")
        break

    ind += 1
else:
    print("Number not found")
```

VII. Wczytuj hasła od użytkownika dopóki nie zostanie wpisane poprawne hasło "python". Jeśli liczba prób przekroczy 3 użyj break. Jeśli użytkownik wpisze poprawne hasło bez przerwania pętli wypisz "Logged in" w bloku else.


```python
password = ""
attempt = 0

while password != "python":
    if attempt >= 3:
        print("Przekroczono ilość prób")
        break

    password = input("Podaj hasło: ")
    attempt += 1
else:
    print("Logged in")
```

VIII. Przeglądaj tekst znak po znaku używając pętli while. Jeśli znajdziesz cyfrę zakończ pętlę używając break. Jeśli pętla zakończy się bez przerwania wypisz "No digits in the text" w bloku else.


```python
text = "jhgfdfghjhgf4fghjkigf456"
ind = 0

while ind < len(text):
    if text[ind].isdigit():
        print("Znaleziono cyfrę")
        break

    ind += 1
else:
    print("No digits in the text")
```

I. Napisz funkcję createList(min_len, max_len), która:
• przyjmuje minimalną i maksymalną długość listy,
• losuje długość z tego zakresu,
• tworzy listę liczb całkowitych od 1 do 100 o tej długości,
• zwraca utworzoną listę.

```python
import random


def create_list(min_len, max_len):
    length = random.randint(min_len, max_len)

    return [random.randint(1, 100) for _ in range(length)]


print(create_list(1, 100))
```
