- przyjmuje listę liczb,

Następnie:

- wygeneruj listę losowych liczb,
- użyj funkcji aby policzyć średnią,
- wyświetl wynik.

```python
import random


def average(numbers):
    return sum(numbers) / len(numbers)


numbers = [1, 2]
print(average(numbers))

numbers_random = [random.randint(1, 10) for _ in range(10)]
print(average(numbers_random))
```

III. Napisz funkcję `count_letters(text)`, która:
- przyjmuje napis,
- zwraca słownik zawierający liczbę wystąpień każdej litery.

Przykład:\
"ala" → {"a":2,"l":1}\
Następnie wyświetl wynik w uporządkowanej formie.

Napisz funkcję `longest_word(words)`, która:
- przyjmuje listę słów,
- zwraca najdłuższe słowo.

Jeśli istnieje kilka najdłuższych słów, zwróć pierwsze z nich.

```python
def count_letters(text):
    dct = {}

    for i in range(len(text)):
        if text[i] not in dct:
            dct[text[i]] = 0
        dct[text[i]] += 1

    dct = dict(sorted(dct.items()))

    return dct


def longest_word(words):
    return max(words, key=len)


text = "ala"
print(count_letters(text))

words = ["a", "aa", "cacv", "aaaa"]
print(longest_word(words))
```

# Funkcje ze słownikami
I. Napisz funkcję `count_names(names)`, która:
- przyjmuje listę imion (mogą się powtarzać),
- zwraca słownik, w którym kluczami są imiona, a wartościami liczba ich wystąpień.

Następnie:
- wyświetl słownik,
- znajdź imię, które występuje najczęściej.

```python
def count_names(names):
    dct = {}

    for i in names:
        if i not in dct:
            dct[i] = 0
        dct[i] += 1

    return dct


names = ["Anna", "Anna", "Anna", "Wojtek"]
names_result = count_names(names)

print(names_result)
print(max(names_result, key=names_result.get))
```

II. Napisz funkcję `invert_dict(d)`, która:
- przyjmuje słownik `d`,
- zwraca nowy słownik, w którym kluczami są wartości z `d`, a wartościami listy
kluczy z oryginalnego słownika.

Przykład:
`{"a":1, "b":2, "c":1}` → `{1:["a","c"], 2:["b"]}`


```python
def invert_dict(d):
    dct = {}

    for key, value in d.items():
        if value not in dct:
            dct[value] = []
        dct[value].append(key)

    return dct


dct = {"a": 1, "b": 2, "c": 1}
print(invert_dict(dct))
```

III. Napisz funkcję `filter_high_scores(scores, threshold)`, która:
- przyjmuje słownik z imionami i wynikami punktowymi,
- przyjmuje wartość progową threshold,
- zwraca nowy słownik z osobami, które mają wynik większy niż threshold.

Następnie:
- oblicz średni wynik wszystkich osób,
- wyświetl słownik osób z wynikami powyżej progu.

```python
def filter_high_scores(scores, threshold):
    scores_above_threshold = {}

    for key, value in scores.items():
        if value > threshold:
            scores_above_threshold[key] = value

    return scores_above_threshold


scores = {"a": 56, "b": 24, "c": 87}
threshold = 50

print(sum(scores.values()) / len(scores))
print(filter_high_scores(scores, threshold))
```

IV. Napisz funkcję `letter_frequency(text)`, która:
- przyjmuje napis,
- zwraca słownik, w którym kluczami są litery, a wartościami liczba ich wystąpień,
- ignoruje wielkość liter (czyli "A" i "a" traktuje jako jedną literę).
Następnie:
- wyświetl litery posortowane malejąco według liczby wystąpień,
- przetestuj funkcję dla kilku różnych napisów.


```python
def letter_frequency(text):
    dct = {}

    for i in text.lower():
        if i not in dct:
            dct[i] = 0
        dct[i] += 1

    return dct


text = "AAaaBva"
result = letter_frequency(text)

print(sorted(result.items(), key=lambda x: x[1], reverse=True))
```

V. Napisz funkcję `tournament_results(results)`, która:
- przyjmuje słownik, gdzie kluczami są zawodnicy, a wartościami liczba zwycięstw,
- zwraca nowy słownik, w którym kluczami są liczby zwycięstw, a wartościami
listy zawodników z taką liczbą zwycięstw,
- zwraca także listę krotek `(liczba zwycięstw, lista zawodników)` posortowaną malejąco według liczby zwycięstw.

Następnie:
- wyświetl słownik i posortowaną listę krotek,
- przetestuj funkcję dla przykładowych danych.

```python
def tournament_results(results):
    dct = {}

    for key, value in results.items():
        if value not in dct:
            dct[value] = []
        dct[value].append(key)

    dct_sorted = sorted(dct.items(), key=lambda item: item[0], reverse=True)

    return dct, dct_sorted


results = {"a": 2, "b": 0, "c": 3, "d": 0, "e": 1}
dct, dct_sorted = tournament_results(results)

print(dct)
print(dct_sorted)
```

# Funkcje z setami
I. Napisz funkcję `unique_names(names)`, która:
- przyjmuje listę imion (mogą się powtarzać),
- zwraca zbiór (`set`) zawierający tylko unikalne imiona.

Następnie:
- wyświetl liczbę unikalnych imion,
- wyświetl wszystkie unikalne imiona w kolejności alfabetycznej.

```python
def unique_names(names):
    return set(names)


names = ["b", "a", "a", "A"]
unique = unique_names(names)

print(len(unique))
print(sorted(list(unique)))
```

II. Napisz funkcję `common_elements(set1, set2)`, która:
- przyjmuje dwa zbiory,
- zwraca zbiór elementów wspólnych (przecięcie zbiorów).

Następnie:
- przetestuj funkcję dla dwóch przykładowych zbiorów,
- wyświetl wynik.

```python
def common_elements(set1, set2):
    return set1.intersection(set2)


lst1 = ["a", "b", "c", "a"]
lst2 = ["a", "b", "d", "a"]
set1 = set(lst1)
set2 = set(lst2)

print(common_elements(set1, set2))
```

III. Napisz funkcję `all_unique(numbers)`, która:
- przyjmuje listę liczb,
- zwraca `True`, jeśli wszystkie liczby są unikalne, `False` w przeciwnym wypadku.

Następnie:
- przetestuj funkcję dla różnych list liczb,
- wyświetl wynik dla każdej listy.

```python
def all_unique(numbers):
    return len(numbers) == len(set(numbers))


lst1 = [1, 2, 3, 4]
lst2 = [1, 1, 2, 1]

print(all_unique(lst1))
print(all_unique(lst2))
```

IV. Napisz funkcję `set_operations(set1, set2)`, która:
- przyjmuje dwa zbiory,
- zwraca słownik z kluczami "union", "intersection", "difference", "symmetric_differenc"
- wartościami słownika są odpowiednie wyniki operacji na zbiorach.

Następnie:
- wyświetl wszystkie wyniki operacji dla przykładowych zbiorów.


```python
def set_operations(set1, set2):
    dct = {}

    dct["union"] = set(set1).union(set(set2))
    dct["intersection"] = set(set1).intersection(set(set2))
    dct["difference"] = set(set1).difference(set(set2))
    dct["symmetric_difference"] = set(set1).symmetric_difference(set(set2))

    return dct


set1 = {1, 2, 3, 4, 5}
set2 = {3, 4, 5, 6, 7}

print(set_operations(set1, set2))
```

V. Napisz funkcję `remove_duplicates(lst)`, która:
- przyjmuje listę z możliwymi powtórzeniami,
- zwraca listę elementów bez powtórzeń, zachowując oryginalną kolejność (hint:
użyj `set` do sprawdzenia unikalności, ale listy do zachowania kolejności).

Następnie:
- przetestuj funkcję dla różnych list,
- wyświetl wyniki.


```python
def remove_duplicates(lst):
    st = set()
    new_lst = []

    for i in lst:
        if i not in st:
            st.add(i)
            new_lst.append(i)

    return new_lst


lst1 = [1, 2, 3, 4, 1, 2, 3, 1, 1]
lst2 = ['f', 'a', 'c', 'e', 'b', 'a', 'f']

print(remove_duplicates(lst1))
print(remove_duplicates(lst2))
```

II. Napisz funkcję `compare_tuple_list(t, l)`, która:
- przyjmuje krotkę `t` i listę `l`,
- zwraca słownik z informacją, które elementy są wspólne, a które różne między
tymi strukturami.

Przykład wyniku:\
`{"common":[...], "t_only":[...], "l_only":[...]}`

```python
def compare_tuple_list(t, l):
    dct = {}
    set_tuple = set(t)
    set_list = set(l)

    dct["common"] = list(set_tuple.intersection(set_list))
    dct["t_only"] = list(set_tuple.difference(set_list))
    dct["l_only"] = list(set_list.difference(set_tuple))

    return dct


t = (1, 2, 3, 4)
l = [3, 4, 5, 6]

print(compare_tuple_list(t, l))
```

III. Napisz funkcję `tuple_immutable(t)`, która:
- przyjmuje krotkę `t`,
- próbuje zmodyfikować pierwszy element krotki i zwraca komunikat, czy modyfikacja jest możliwa,
- przetestuj funkcję również dla listy `l` o tych samych elementach.

```python
def tuple_immutable(t):
    try:
        t[0] = "a"
        return "Modyfikacja powiodła się"
    except TypeError:
        return "Modyfikacja nie powiodła się"


t = (1, 2, 3)
l = [1, 2, 3]

print(tuple_immutable(t))
print(tuple_immutable(l))
```

# Funkcje z *args i **kwargs
I. Napisz funkcję `sum_all(*args)`, która:

Następnie:
- przetestuj funkcję dla różnych zestawów liczb, np. `sum_all(1,2,3)`, `sum_all(10,20,30,40)`,
- wyświetl wynik dla każdej próby.

II. Napisz funkcję `multiply_all(*args)`, która:
- przyjmuje dowolną liczbę argumentów liczbowych,
- zwraca ich iloczyn,
- jeśli nie podano żadnego argumentu, zwraca 1.

Przetestuj funkcję dla różnych zestawów liczb.


```python
def multiply_all(*args):
    result = 1

    for arg in args:
        result *= arg

    return result


print(multiply_all(1, 2, 3))
print(multiply_all())
```

IV. Napisz funkcję `describe_person(*args, **kwargs)`, która:
- przyjmuje dowolną liczbę argumentów pozycyjnych reprezentujących cechy (np.
`wysoki`, `inteligentny`),
- przyjmuje dowolną liczbę argumentów nazwanych opisujących dane osoby (np.
`name="Jan"`, `age=30`),
- wypisuje najpierw cechy w jednej linii, a potem dane w formie `klucz: wartość`.

Przetestuj funkcję dla różnych zestawów argumentów.

```python
def describe_person(*args, **kwargs):
    print("Cechy:", *args)

    for key, value in kwargs.items():
        print(f"{key}: {value}")


describe_person("wysoki", "inteligentny", name="Jan", age=22)
```

V. Napisz funkcję `stats_calculator(*args, **kwargs)`, która:
- przyjmuje dowolną liczbę liczb przez `*args` i oblicza ich średnią oraz sumę,
- przyjmuje dodatkowo parametry nazwane przez `**kwargs`, np. `round_digits=2`,
który określa liczbę miejsc po przecinku w wynikach,
- zwraca słownik z kluczami: `sum`, `average`.

Przetestuj funkcję dla różnych zestawów danych liczbowych i wartości `round_digits`.

```python
def stats_calculator(*args, **kwargs):
    dct = {}
    total_sum = sum(args)
    avg = total_sum / len(args)

    r = kwargs.get("round_digits", 1)

    dct["sum"] = round(total_sum, r)
    dct["average"] = round(avg, r)

    return dct


print(stats_calculator(1, 2, 3))
print(stats_calculator(1.26, 2.46, 3.93, round_digits=1))
print(stats_calculator(13.513, 2.141, 3.125, round_digits=2))
```
